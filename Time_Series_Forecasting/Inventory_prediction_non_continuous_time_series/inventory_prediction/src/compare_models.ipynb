```python
def run_experiment(df,output_col,n_best):
    output_path = os.path.join(best_models_folder, f'tx1_{output_col}_amount_sold_best_models.csv')
    exp = TSForecastingExperiment()
    exp.setup(
        data=df,
        target=output_col,
        fold_strategy='sliding',
        fold=2,
        fh=14,
        session_id=123,
        numeric_imputation_target='linear',
    )

    best_n_models = exp.compare_models(sort='RMSE',
                            turbo=True,
                            n_select=5,
                            budget_time=5)

    print(f"Best {n_best} models for {output_col}:")

    return
```

## Run experiments to compare best models


```python
#find best models for tx1_FOODS_3_586
best_models_folder = '/home/leon/Documents/github_repos/bd_take_home_test/inventory_prediction/reports/best_models'
if not os.path.exists(best_models_folder):
    os.makedirs(best_models_folder)

output_path = os.path.join(best_models_folder, f'tx1_FOODS_3_586_best_models.csv')

run_experiment(train_subset_586, 'FOODS_3_586_amount_sold', 5, output_path)
```
